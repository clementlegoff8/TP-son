# **TP - Un peu de musique**

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio,display
from scipy.io import wavfile

RATE = 44_100
LA = 440
DO = 523.25
```

## 2.1 Génération d'une sinusoïde

1. Il faut 44100 échantillons dans le tableau pour une durée de 1 seconde (de manière générale, le nombre d'échantillons nécessaires correspond au RATE multiplié par la durée).

2. La position de la membrane en fonction du temps pour une fréquence phi est donnée par $f(t)=sin(2\pi\phi t)$.

3.

```python
temps = np.linspace(0,1,44100) #liste des temps
position = np.sin(np.multiply(temps,2*np.pi*440)) #position de la membrane
plt.plot(temps[:int(44100*0.05)],position[:int(44100*0.05)]) #affichage des 50 premières ms
plt.show()
display(Audio(position,rate=RATE)) #pour écouter le son
```

4. Généralisation du code :

```python
def display_signal(signal,rate=RATE,title="Signal",width=1):
    temps = np.linspace(0,1,rate)
    plt.plot(temps[:int(44100*width)],signal[:int(44100*width)])
    plt.title(title)
    plt.show()
    display(Audio(signal,rate=rate))

display_signal(position,width=0.05)
```

## 2.2 Généralisation : la fonction `sine()`

```python
def sine(freq,duration=1,amplitude=1.): #renvoie un signal sinusoïdal à la fréquence freq de durée duration et d'amplitude amplitude
    temps = np.linspace(0,duration,RATE*duration)
    signal = amplitude * np.sin(np.multiply(temps,2*np.pi*freq))
    return signal

display_signal(sine(300),width=0.05) #test à la fréquence 300 Hz
```

## 2.3 Pour aller plus loin : glissando (Effet "note qui monte")

```python
def sine_linear(freq1,freq2,duration=1): #renvoie un glissando de freq1 à freq2 de durée duration
    frequence = np.linspace(freq1,freq2,RATE*duration) #fréquence linéairement variable
    signal = np.sin(2*np.pi*frequence.cumsum()/RATE) #formule générale d'un signal de fréquence variable (cumsum = somme cumulative, et en divisant par RATE on se retrouve avec une valeur approchée de l'intégrale de la fréquence)
    return signal

display_signal(sine_linear(440,523.25),width=0.05)
```

## 3.1 Création d'un crescendo

1. On peut refaire un np.linspace pour l'amplitude.

2. et 3.

```python
def crescendo_sine(freq,duration=1,increase=True):
    temps = np.linspace(0,duration,RATE*duration)
    if increase:
        amp = np.linspace(0,1,RATE*duration) #crescendo
    else:
        amp = np.linspace(1,0,RATE*duration) #decrescendo
    signal = np.multiply(amp,np.sin(np.multiply(temps,2*np.pi*freq)))
    return signal

display_signal(crescendo_sine(440,increase=False))
```

## 4. Concaténation de sons

```python
temps = np.linspace(0,1,RATE)
la = np.sin(np.multiply(temps,2*np.pi*440))
do = np.sin(np.multiply(temps,2*np.pi*523.25))
succession = np.concatenate((la,do)) #concaténation du la et du do
display(Audio(succession,rate=RATE))
```

## 5.3 Mise à l'échelle

```python
def float_to_int16(as_float): #convertit un tableau de flottants entre -1 et 1 en tableau d'entiers int16
    return (as_float * 32767).astype(np.int16)
```

## 6.2 Calcul des rapports

```python
ratios = np.array([(2**(i/12)) for i in range(13)]) #tableau des 13 rapports entre les notes de la gamme
notes = {"do":0,"do#":1,"ré":2,"ré#":3,"mi":4,"fa":5,"fa#":6,"sol":7,"sol#":8,"la":9,"la#":10,"si":11} #indexation des notes
def freq_from_name(name): #renvoie la fréquence d'une note à partir de son nom
    i = notes[name]
    return 440*2**((i-9)/12) # 9 est l'index du la

print(freq_from_name("la"))
```

## 7. Les accords : Superposition de sons

Il faut additionner les tableaux (s'ils sont bien de la même longueur). Exemple avec l'accord de do (do mi sol) :

```python
temps = np.linspace(0,1,RATE)
do = np.sin(np.multiply(temps,2*np.pi*freq_from_name("do")))
mi = np.sin(np.multiply(temps,2*np.pi*freq_from_name("mi")))
sol = np.sin(np.multiply(temps,2*np.pi*freq_from_name("sol")))
accord = do + mi + sol #création de l'accord en additionnant terme à terme les éléments des 3 tableaux
display(Audio(accord,rate=RATE))
```

## 9. Analyse d'un son réel

```python
samplerate,data = wavfile.read("media/sounds-cello.wav") #lecture du fichier sounds-cello.wav
display(Audio(data,rate=rate))
print(f"Le taux d'échantillonnage est {samplerate}.") #affichage du taux d'échantillonnage
n = len(data) #nombre total d'échantillons
print(f"Le nombre total d'échantillons est {n}.")
duree = n/rate #durée du morceau
print(f"La durée du morceau est {duree} secondes.")
plt.plot(np.linspace(0,duree,n),data) #affichage du signal
plt.show()
```

## 10. Ajout d'un effet d'écho

Exercice v1

```python
# Paramètres pour l'écho
delay = 2  # en secondes
main_ratio, delayed_ratio = 0.7, 0.3

offset = RATE*delay #conversion du delay en nombre d'échantillons
echo = np.zeros(n) #tableau de 0 de la bonne taille
echo[offset:]=data[:(n-offset)] #on rajoute le signal original mais décalé
son_avec_echo = main_ratio*data + delayed_ratio*echo #on ajuste les ratios
plt.plot(np.linspace(0,duree,n),son_avec_echo) #affichage du signal
display(Audio(son_avec_echo,rate=rate))
```

Exercice v2 (durée plus longue)

```python
# Paramètres pour l'écho
delay = 2  # en secondes
main_ratio, delayed_ratio = 0.7, 0.3

offset = RATE*delay #conversion du delay en nombre d'échantillons
original_allonge = np.concatenate(data,np.array([0 for i in range(offset)]))
echo = np.zeros(n+offset) #tableau de 0 de la bonne taille
echo[offset:]=data[:] #on rajoute le signal original mais décalé
son_avec_echo = main_ratio*original_allonge + delayed_ratio*echo #on ajuste les ratios
plt.plot(np.linspace(0,duree,n),son_avec_echo) #affichage du signal
display(Audio(son_avec_echo,rate=rate))
```
